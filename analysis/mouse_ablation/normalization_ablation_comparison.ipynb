```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
from pathlib import Path

# ---------------------------------------------------------------------------
# 1. Configuration
# ---------------------------------------------------------------------------
ablation_timestamp_pairs = [
    # --- No norm rank ---
    ("stjepa_v2_no_norm",                                          "21032026_085513_196"),
    ("stjepa_v2_shifted_log",                                      "24032026_224828_432"),
    ("stjepa_v2_read_depth_shifted_log",                           "20032026_215239_257"),
    ("stjepa_v2_gene_corrected_read_depth_shifted_log",            "21032026_213950_914"),
    ("stjepa_v2_mean_no_norm_rank",                                "23032026_204349_268"),
    ("stjepa_v2_nz_mean_no_norm_rank",                             "25032026_010748_327"),
    ("stjepa_v2_gene_corrected_read_depth_mean_no_norm_rank",      "25032026_031148_546"),
    ("stjepa_v2_gene_corrected_read_depth_nz_mean_no_norm_rank",   "25032026_052014_360"),
    ("stjepa_v2_seurat_no_norm_rank",                              "22032026_135514_292"),
    ("stjepa_v2_pearson_residuals_no_norm_rank",                   "25032026_071947_719"),
    # --- Norm rank ---
    ("stjepa_v2_mean",                                             "24032026_022512_769"),
    ("stjepa_v2_nz_mean",                                          "24032026_040014_694"),
    ("stjepa_v2_gene_corrected_read_depth_mean",                   "21032026_190736_759"),
    ("stjepa_v2_gene_corrected_read_depth_nz_mean",                "21032026_182612_890"),
    ("stjepa_v2_seurat",                                           "21032026_154941_529"),
    ("stjepa_v2_pearson_residuals",                                "21032026_172808_828"),
    ("stjepa_v2_no_norm_gene_corrected_read_depth_mean_rank",      "23032026_233225_877"),
    ("stjepa_v2_no_norm_gene_corrected_read_depth_nz_mean_rank",   "24032026_005846_090"),
    ("stjepa_v2_no_norm_mean_rank",                                "23032026_220800_207"),
    ("stjepa_v2_no_norm_nz_mean_rank",                             "23032026_192023_653"),
    ("stjepa_v2_no_norm_seurat_rank",                              "08042026_193915_019"),
    ("stjepa_v2_no_norm_pearson_residuals_rank",                   "08042026_201620_913"),
]

ABLATION_LABELS = {
    # No norm rank
    "stjepa_v2_no_norm":                                          "No Normalization",
    "stjepa_v2_shifted_log":                                      "Shifted Log",
    "stjepa_v2_read_depth_shifted_log":                           "Read Depth Shifted Log",
    "stjepa_v2_gene_corrected_read_depth_shifted_log":            "Gene Corr. RD Shifted Log",
    "stjepa_v2_mean_no_norm_rank":                                "Mean + No Norm Rank",
    "stjepa_v2_nz_mean_no_norm_rank":                             "NZ Mean + No Norm Rank",
    "stjepa_v2_gene_corrected_read_depth_mean_no_norm_rank":      "Gene Corr. RD Mean + No Norm Rank",
    "stjepa_v2_gene_corrected_read_depth_nz_mean_no_norm_rank":   "Gene Corr. RD NZ Mean + No Norm Rank",
    "stjepa_v2_seurat_no_norm_rank":                              "Seurat v3 + No Norm Rank",
    "stjepa_v2_pearson_residuals_no_norm_rank":                   "Pearson Residuals + No Norm Rank",
    # Norm rank
    "stjepa_v2_mean":                                             "Mean",
    "stjepa_v2_nz_mean":                                          "Nonzero Mean",
    "stjepa_v2_gene_corrected_read_depth_mean":                   "Gene Corr. RD Mean",
    "stjepa_v2_gene_corrected_read_depth_nz_mean":                "Gene Corr. RD Nonzero Mean",
    "stjepa_v2_seurat":                                           "Seurat v3",
    "stjepa_v2_pearson_residuals":                                "Pearson Residuals",
    "stjepa_v2_no_norm_gene_corrected_read_depth_mean_rank":      "No Norm + Gene Corr. RD Mean Rank",
    "stjepa_v2_no_norm_gene_corrected_read_depth_nz_mean_rank":   "No Norm + Gene Corr. RD NZ Mean Rank",
    "stjepa_v2_no_norm_mean_rank":                                "No Norm + Mean Rank",
    "stjepa_v2_no_norm_nz_mean_rank":                             "No Norm + NZ Mean Rank",
    "stjepa_v2_no_norm_seurat_rank":                              "No Norm + Seurat v3 Rank",
    "stjepa_v2_no_norm_pearson_residuals_rank":                   "No Norm + Pearson Residuals Rank",
}

# Group membership for colour coding
NO_NORM_RANK_ABLATIONS = {a for a, _ in ablation_timestamp_pairs[:10]}
NORM_RANK_ABLATIONS = {a for a, _ in ablation_timestamp_pairs[10:]}

base_path = Path("/nfs/team361/sb75/stjepa-reproducibility/artifacts/ablation")
results_dir = base_path / "results"
figures_dir = base_path / "figures"
results_dir.mkdir(parents=True, exist_ok=True)
figures_dir.mkdir(parents=True, exist_ok=True)

# Panel titles and ordering
REGION_TITLES = {
    "ccf_region_name":             "Tissue Regions (MERFISH)",
    "Sub_molecular_tissue_region": "Tissue Regions (STARmap)",
}
REGION_ORDER = ["ccf_region_name", "Sub_molecular_tissue_region"]

# ---------------------------------------------------------------------------
# 2. Load niche identification CSVs
# ---------------------------------------------------------------------------
frames = []
for ablation, timestamp in ablation_timestamp_pairs:
    fpath = (
        base_path / ablation / timestamp
        / "mmb0-1b_smb1-1b_1p_neighborhood_emb_niche_identification_metrics_res_sweep.csv"
    )
    if not fpath.exists():
        print(f"[WARN] File not found, skipping: {fpath}")
        continue
    df = pd.read_csv(fpath)
    df["ablation"] = ablation
    df["model_timestamp"] = timestamp
    frames.append(df)

if not frames:
    raise FileNotFoundError("No niche identification CSV files were found.")

df_all = pd.concat(frames, ignore_index=True)

print("=== Data points per (ablation, region_key) ===")
print(df_all.groupby(["ablation", "region_key"]).size().unstack(fill_value=0))
print()

df_all.to_csv(results_dir / "niche_identification_normalization_ablations.csv", index=False)
print(f"Saved to {results_dir / 'niche_identification_normalization_ablations.csv'}\n")

# ---------------------------------------------------------------------------
# 3. Style settings
# ---------------------------------------------------------------------------
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
    "font.size": 9,
    "axes.labelsize": 10,
    "axes.titlesize": 11,
    "xtick.labelsize": 8.5,
    "ytick.labelsize": 8.5,
    "legend.fontsize": 8.5,
    "figure.dpi": 300,
    "savefig.dpi": 300,
    "savefig.bbox": "tight",
    "savefig.pad_inches": 0.05,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
})

# Two-tone blue palette: dark blue for No Norm Rank, light blue for Norm Rank
COLOR_NO_NORM_RANK = "#08519C"
COLOR_NORM_RANK = "#9ECAE1"

def get_ablation_color(ablation: str) -> str:
    if ablation in NO_NORM_RANK_ABLATIONS:
        return COLOR_NO_NORM_RANK
    return COLOR_NORM_RANK

# ---------------------------------------------------------------------------
# 4. Horizontal bar plot for niche identification
# ---------------------------------------------------------------------------
def plot_niche_horizontal(
    df: pd.DataFrame,
    metric: str,
    ablation_order: list[str],
    higher_is_better: bool = True,
):
    """
    Horizontal bar plot of mean metric per ablation with individual points.
    One panel per region_key, side by side.
    """
    region_keys = REGION_ORDER
    n_panels = len(region_keys)
    n_ab = len(ablation_order)
    row_height = 0.35

    fig, axes = plt.subplots(
        1, n_panels,
        figsize=(3.5 * n_panels, row_height * n_ab + 1.0),
        sharex=True,
        constrained_layout=True,
    )
    if n_panels == 1:
        axes = [axes]

    # Fixed colour per ablation based on group
    for ax, rk in zip(axes, region_keys):
        sub = df[df["region_key"] == rk]

        # Compute means and sort ascending (so highest ends up at top)
        mean_map = {}
        for a in ablation_order:
            vals = sub.loc[sub["ablation"] == a, metric].values
            mean_map[a] = np.mean(vals) if len(vals) > 0 else 0

        sorted_ablations = sorted(
            ablation_order,
            key=lambda a: mean_map[a],
            reverse=not higher_is_better,  # ascending for higher-is-better (best on top), descending for lower-is-better
        )

        means, sds, all_vals = [], [], []
        for a in sorted_ablations:
            vals = sub.loc[sub["ablation"] == a, metric].values
            means.append(np.mean(vals) if len(vals) > 0 else 0)
            sds.append(np.std(vals) if len(vals) > 0 else 0)
            all_vals.append(vals)

        y = np.arange(n_ab)
        bar_colors = [get_ablation_color(a) for a in sorted_ablations]

        # --- bars (mean) ---
        ax.barh(
            y, means, height=0.55,
            color=bar_colors, edgecolor="white", linewidth=0.6,
            alpha=0.80, zorder=2,
        )

        # --- error bars (± 1 SD) ---
        ax.errorbar(
            means, y, xerr=sds,
            fmt="none", ecolor="0.25", elinewidth=0.8,
            capsize=2.0, capthick=0.8, zorder=3,
        )

        # --- individual data points (jittered strip) ---
        rng = np.random.default_rng(42)
        for i, vals in enumerate(all_vals):
            n_pts = len(vals)
            if n_pts == 0:
                continue
            jitter = rng.uniform(-0.15, 0.15, size=n_pts)
            ax.scatter(
                vals, np.full(n_pts, i) + jitter,
                s=10, color="0.15", edgecolor="white",
                linewidth=0.3, zorder=4, alpha=0.60,
            )

        # --- formatting ---
        pretty_rk = REGION_TITLES.get(rk, rk.replace("_", " "))
        ax.set_title(pretty_rk, fontweight="medium")
        ax.set_yticks(y)
        ax.set_yticklabels(
            [ABLATION_LABELS.get(a, a) for a in sorted_ablations],
            fontsize=7.5,
        )
        ax.set_xlabel(f"{metric} (\u2191)")
        ax.set_ylim(-0.5, n_ab - 0.5)
        ax.xaxis.set_major_locator(mpl.ticker.MaxNLocator(5))
        ax.grid(axis="x", linewidth=0.3, alpha=0.5)

    return fig


# ---------------------------------------------------------------------------
# 5. Generate and save niche identification figures
# ---------------------------------------------------------------------------
ablation_order = [a for a, _ in ablation_timestamp_pairs]

for metric in ["NMI", "ARI"]:
    fig = plot_niche_horizontal(df_all, metric=metric, ablation_order=ablation_order, higher_is_better=True)
    for ext in ["svg", "pdf", "png"]:
        out = figures_dir / f"ablation_norm_niche_{metric}_comparison.{ext}"
        fig.savefig(out)
    print(f"Saved ablation_norm_niche_{metric}_comparison.svg / .pdf / .png")

plt.show()

# ===========================================================================
# 6. Batch integration metrics
# ===========================================================================
batch_frames = []
for ablation, timestamp in ablation_timestamp_pairs:
    fpath = (
        base_path / ablation / timestamp
        / "mmb0-1b_smb1-1b_1p_neighborhood_emb_batch_integration_metrics.csv"
    )
    if not fpath.exists():
        print(f"[WARN] Batch file not found, skipping: {fpath}")
        continue
    df_batch = pd.read_csv(fpath)
    df_batch["ablation"] = ablation
    df_batch["model_timestamp"] = timestamp
    batch_frames.append(df_batch)

if not batch_frames:
    raise FileNotFoundError("No batch integration CSV files were found.")

df_batch_all = pd.concat(batch_frames, ignore_index=True)

df_batch_all.to_csv(
    results_dir / "batch_integration_normalization_ablations.csv", index=False
)
print(f"\nSaved batch integration dataframe to "
      f"{results_dir / 'batch_integration_normalization_ablations.csv'}")

df_batch_wide = df_batch_all.pivot_table(
    index="ablation", columns="metric", values="score"
).reset_index()
print("\n=== Batch integration metrics ===")
print(df_batch_wide.to_string(index=False))

# ---------------------------------------------------------------------------
# 6b. Horizontal bar chart per batch metric
# ---------------------------------------------------------------------------
BATCH_METRIC_DIRECTION = {
    "iLISI": "\u2191",
    "ASW":   "\u2193",
    "MMD":   "\u2193",
}

ablation_order_map_unused = None  # colours now handled by get_ablation_color
batch_metrics = df_batch_all["metric"].unique()

for bm in batch_metrics:
    # Determine sort direction: best on top
    is_higher_better = BATCH_METRIC_DIRECTION.get(bm, "") == "\u2191"
    df_sub = (
        df_batch_all[df_batch_all["metric"] == bm][["ablation", "score"]]
        .sort_values("score", ascending=is_higher_better)
        .reset_index(drop=True)
    )

    fig, ax = plt.subplots(
        figsize=(3.5, 0.3 * len(df_sub) + 0.6),
        constrained_layout=True,
    )

    labels = [ABLATION_LABELS.get(a, a) for a in df_sub["ablation"]]
    values = df_sub["score"].astype(float).values
    y = np.arange(len(labels))

    bar_colors = [get_ablation_color(a) for a in df_sub["ablation"]]

    ax.barh(
        y, values, height=0.55,
        color=bar_colors, edgecolor="white", linewidth=0.6,
        alpha=0.85, zorder=2,
    )

    x_max = float(np.max(values)) if len(values) > 0 else 1.0
    for yi, v in zip(y, values):
        ax.text(
            v + x_max * 0.015, yi, f"{v:.4f}",
            va="center", ha="left", fontsize=7,
        )

    ax.set_yticks(y)
    ax.set_yticklabels(labels, fontsize=7.5)
    direction = BATCH_METRIC_DIRECTION.get(bm, "")
    ax.set_xlabel(f"{bm} ({direction})" if direction else bm)
    ax.set_title(f"Batch Integration – {bm}", fontweight="medium")
    ax.set_xlim(0, x_max * 1.15)
    ax.xaxis.set_major_locator(mpl.ticker.MaxNLocator(5))
    ax.grid(axis="x", linewidth=0.3, alpha=0.5)

    for ext in ["svg", "pdf", "png"]:
        out = figures_dir / f"ablation_norm_batch_integration_{bm}.{ext}"
        fig.savefig(out)
    print(f"Saved ablation_norm_batch_integration_{bm}.svg / .pdf / .png")

plt.show()
```

### 1.5 Define Functions

```python
def plot_metric_comparison(
    df: pd.DataFrame,
    metric: str,
    ablation_order: list[str],
    figsize_per_panel: tuple = (3.0, 2.4),
):
    """
    Bar plot of mean metric per ablation with individual data points overlaid.
    One panel per region_key. Text exported as editable paths for Illustrator.
    """
    region_keys = REGION_ORDER
    n_panels = len(region_keys)
    fig, axes = plt.subplots(
        1, n_panels,
        figsize=(figsize_per_panel[0] * n_panels, figsize_per_panel[1]),
        sharey=True,
        constrained_layout=True,
    )
    if n_panels == 1:
        axes = [axes]

    n_ablations = len(ablation_order)
    bar_width = 0.6
    colors = PALETTE[:n_ablations]

    for ax, rk in zip(axes, region_keys):
        sub = df[df["region_key"] == rk]

        means, sds, all_vals = [], [], []
        for a in ablation_order:
            vals = sub.loc[sub["ablation"] == a, metric].values
            means.append(np.mean(vals) if len(vals) > 0 else 0)
            sds.append(np.std(vals) if len(vals) > 0 else 0)
            all_vals.append(vals)

        x = np.arange(n_ablations)

        # --- bars (mean) ---
        ax.bar(
            x, means, width=bar_width,
            color=colors, edgecolor="white", linewidth=0.6,
            alpha=0.75, zorder=2,
        )

        # --- error bars (± 1 SD) ---
        ax.errorbar(
            x, means, yerr=sds,
            fmt="none", ecolor="0.25", elinewidth=0.8,
            capsize=2.5, capthick=0.8, zorder=3,
        )

        # --- individual data points (jittered strip) ---
        rng = np.random.default_rng(42)
        for i, vals in enumerate(all_vals):
            n_pts = len(vals)
            if n_pts == 0:
                continue
            jitter = rng.uniform(-0.15, 0.15, size=n_pts)
            ax.scatter(
                np.full(n_pts, i) + jitter, vals,
                s=12, color="0.15", edgecolor="white",
                linewidth=0.3, zorder=4, alpha=0.65,
            )

        # --- formatting ---
        pretty_rk = REGION_TITLES.get(rk, rk.replace("_", " "))
        ax.set_title(pretty_rk, fontweight="medium")
        ax.set_xticks(x)
        ax.set_xticklabels(
            [ABLATION_LABELS.get(a, a) for a in ablation_order],
            rotation=45, ha="right",
        )
        ax.set_ylabel(f"{metric} (\u2191)" if ax is axes[0] else "")
        ax.set_xlim(-0.5, n_ablations - 0.5)
        ax.yaxis.set_major_locator(mpl.ticker.MaxNLocator(5))
        ax.grid(axis="y", linewidth=0.3, alpha=0.5)

    return fig
```
