```python
import pickle
import re
import pprint
```

```python
token_dictionary_file_path = '/lustre/scratch126/cellgen/lotfollahi/DATASETS/tokenizer/token_dictionary_mus_musculus.pkl'
token_dictionary_file_path = '/nfs/team361/sb75/DATASETS/tokenizer/token_dictionary_mus_musculus.pkl'
```

```python
# Load token dict
with open(token_dictionary_file_path, 'rb') as f:
    token_dict = pickle.load(f)

token_dict_reverse =  {value: key for key, value in token_dict.items()}

spv_dict = {
    k: v for k, v in token_dict.items() if k.startswith('spv_')}
spv_start_idx = min(spv_dict.values())
max_cls_tokens = sum(1 for key in token_dict if "cls" in key)
spv_idx_subtract = spv_start_idx - 2 - max_cls_tokens
```

```python
dummy_indices = []

for value in token_dict.keys():
    match = re.match(r'spv_dummy_(\d+)', str(value))
    if match:
        dummy_indices.append(int(match.group(1)))

if dummy_indices:
    lowest_index = min(dummy_indices)
    print("Lowest dummy index:", lowest_index)
else:
    print("No dummy values found.")
```

```python
# Print token dict
pp = pprint.PrettyPrinter(indent=4, width=40, depth=2)
pp.pprint(token_dict)
```
